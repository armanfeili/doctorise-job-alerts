```python
import logging
import asyncio
from telegram import Update
from telegram.ext import Application

# Set up logging
logging.basicConfig(
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    level=logging.INFO
)

# Function to send an initial message to the channel and log chat ID
async def send_initial_message(application):
    # Send an initial message to the channel to trigger chat ID retrieval
    chat_id = '@DoctoRiseJobAlerts'  # Replace with your channel username
    message = await application.bot.send_message(chat_id=chat_id, text="Hello from the bot!")

    # Log the chat ID of the channel
    actual_chat_id = message.chat_id
    logging.info(f"Channel Chat ID: {actual_chat_id}")
    print(f"Channel Chat ID: {actual_chat_id}")

async def run_bot():
    # Replace 'YOUR_TELEGRAM_BOT_TOKEN' with your bot token
    bot_token = '7235372539:AAFFypcvWhsepq2xLOdB5Y8PjkULuCKaNmc'

    # Create the Application
    application = Application.builder().token(bot_token).build()

    # Initialize the bot (new requirement in version 20.x)
    await application.initialize()

    # Start the bot and send an initial message
    await application.start()
    await send_initial_message(application)

    # Keep the bot running for a while to capture messages
    try:
        await asyncio.sleep(10)  # Adjust the sleep time as necessary
    finally:
        await application.stop()

# Run the bot using asyncio event loop in Jupyter Lab
await run_bot()
```
